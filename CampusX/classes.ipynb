```python
#Rules: Variables will be kept inside the constructor

#Constructor looks alike:
# def __init__(self)
```

```python
    #any variable we want to create, it will  be created inside the class inside the constructor.
    #and the name of the variable will be written as self.variable
class Atm:   

# constructor -> a function but special inside a class
# bkz to execute the code of this function ("constructor"), we do not need to call this function. this is super power.
# so whenever, an object is created of the class, the code of constructor function is executed itseld.
    def __init__(self):
        self.pin = ''
        self.balance = 0
        # print("main toh execute ho gyaa")
        self.menu()

    def menu(self):
        user_input = input('''
            Hi, How can i help you?
            1. press 1 to create Pin
            2. press 2 to change pin
            3. press 3 to check balance
            4. press 4 to withdraw
            5. Anything else to exit
            ''')


        if user_input == '1':
            #create pin
            self.create_pin()
        elif user_input == '2':
            #change pin
            self.change_pin()
        elif user_input == '3':
            # check balance
            self.check_balance()
        elif user_input == '4':
            # withdraw
            self.withdraw()
        else:
            # exit()
            pass

    def create_pin(self):
        user_pin = input("enter your pin")
        self.pin = user_pin

        # lets ask for the balance of the user
        user_balance = int(input("enter your balance"))
        self.balance = user_balance

        print("pin created successfully")
        self.menu()  # added to ensure continuity

    def change_pin(self):
        # ask for prev pin
        prev_pin = input("enter your previous PIN")
        if self.pin == prev_pin: # if same
            # enter new pin
            new_pin = input("enter your new pin")
            confirm_pin = input("re-enter your new pin")
            if new_pin == confirm_pin:
                self.pin = new_pin
                print("pin changed successfully")
                self.menu() # added to ensure continuity
            else:
                print("pin is mismatched\n Try again once.")
                # self.change_pin()
                self.menu() # added to ensure continuity
        else:
            print("Wrong pin !\n Try Again")
            self.menu() # added to ensure continuity


    def check_balance(self):
        # phle pin sahi btao
        pin = input("enter your pin")
        if self.pin == pin:            
            print("Balance", self.balance)
            self.menu()
        else:
            print("wrong pin!")
            self.menu()


    def withdraw(self):
        # phle pin btao
        pin = input("enter your pin")
        if self.pin == pin:
            # enter the amount you want to withdraw
            amount = int(input("enter the amount"))
            if self.balance >= amount:
                print("Successfully withdrawn")
                self.balance = self.balance - amount
                self.menu() # for continuity
            else:
                print("Insufficient funds")
                self.menu()
        else:
            print("wrong pin!")
            self.menu()


        

    
```

### Fraction class
* lets create a new data type "Fraction data type"

```python
3/4 * 1/2  # i want answer to be fraction not float or decimal.
```

What things do i need to create a fraction class ?

*  i need numerator, denominator

```python
class Fraction():

    def __init__(self,x,y):  # parameterized constructor
        self.num = x
        self.den = y

# if anyone wants to see how the object looks alike, __str__ is called for that
# code should be inside __str__
    def __str__(self):  # also a magic method # -> 
        # return 'idhar'  # 
        return '{}/{}'.format(self.num, self.den)



    # automatically triggers when you put "+" operator b/w two objects of the class
    def __add__(self,other):  # self and other are 2 objects # both have numtr and denomtr 
        # print('idhar')
        
        new_num = self.num * other.den + self.den * other.num
        new_den = self.den * other.den
        return '{}/{}'.format(new_num,new_den)

    def __mul__(self, other):
        new_num = self.num * other.num
        new_den = self.den * other.den
        return '{}/{}'.format(new_num, new_den)

    def __sub__(self, other):
        new_num = self.num * other.den - self.den * other.num
        new_den = self.den * other.den
        return '{}/{}'.format(new_num,new_den)

    def __truediv__(self, other):
        new_num = self.num * other.den
        new_den = self.den * other.num
        return '{}/{}'.format(new_num,new_den)

    def convert_to_decimal(self):
        return self.num / self.den
        

    
        
        
    
```
