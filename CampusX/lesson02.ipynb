### Write OOP classes to handle the following scenarios:
* A user can create and view 2D coordinates
* A user can find out the distance between 2 coordinates
* A user can find find the distance of a coordinate from origin
* A  user can check if a point lies on a given line
* A user can find the distance between a given 2D point and a given line

```python
class Point():
    def __init__(self, a, b):
        self.x = a
        self.y = b
# agar object ko dekhna hai ki kaisa dikhta hai toh use magic method __str__

    def __str__(self):
        return "<{},{}>".format(self.x, self.y)
    def view(self):
        print(tuple((self.x, self.y)))
    def distance(self,other): # self = (x1, y1) other -> (x2,y2)
        # self bhi ek object toh uska bhi x-cord hoga aur y -coordinate hoga
        dist = ((other.x - self.x)**2 + (other.y - self.y)**2)**0.5
        print(dist)
    def distance_from_origin(self):
        dist = ((self.x)**2 + (self.y)**2)**0.5
        print(dist)

        # other way
        # dist = self.distance(CoordinateSytem(0,0))  # creating class ka object inside a class
class Line():
    def __init__(self,a,b,c):
        self.a = a
        self.b = b
        self.c = c
     # lets check how the object looks
    def __str__(self):
        return "{}X + {}Y + {}".format(self.a, self.b, self.c)

    def point_on_line(line,point):  # self ko line boldiya
        if line.a * point.x + line.b * point.y + line.c == 0:
            print("point lies on the line")
        else:
            print("point does not lie on the line")
    def distance_between_point_and_line(line, point):
        num = abs(line.a * point.x + line.b * point.y + line.c)
        den = (line.a**2 + line.b**2)**0.5
        print(num/den)

    def Intersect(line1, line2):
        # if two lines have different slopes, they intersect each other
        # slope = -A/ B
        slope1 = -1 * (line1.a / line1.b)
        slope2 = -1 * (line2.a / line2.b)
        if slope1 == slope2:
            print("given two lines intersect each other")
        else:
            print("lines do not intersect each other")

    
```

#### Reference Variables

- Reference variables hold the objects
- We can create objects without reference variable as well
- An object can have multiple reference variables
- Assigning a new reference variable to an existing object does not create a new object

```python
class Person:
  def __init__(self):
    self.name = 'nitish'
    self.gender = 'male'


```

```python
Person() # calling the class iteslf creates an object

p = Person() # means now we are keeping the memory address of the object in the variable "p"

# p is not the object
# p is a reference variable containing the address of the object

q = p

print(id(q))
print(id(p))

# that means we can give an object to muliple variables.
```

```python
# change attribute value with the help of 2nd object
q.name = 'mohit'
print(q.name)
print(p.name)

# since both are pointing to same adrress.
# changing via one variable will impact overall .
```
